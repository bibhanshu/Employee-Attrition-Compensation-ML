## What is *args

*args stores the data in tuples

Sometimes you don't know how many arguments a function will receive. *args lets a function accept a variable number of positional arguments.

````python
def calculate_total(*args):
    return sum(args)

print(calculate_total(100, 200))
print(calculate_total(100, 200, 300, 400))
````

````python
def calculate_sales(*args):
  Tot_sum = sum(args)
  return Tot_sum

print(calculate_sales(100,200))
print(calculate_sales(100,200,300))
````

**kwargs — it lets a function accept a variable number of keyword arguments, which are collected into a dictionary.



````python
def employee_details(**kwargs):
    print(kwargs)

employee_details(name="Rahul", salary=60000, department="Analytics")
````

````python
def employee_details(**kwargs):
  return kwargs

print(employee_details(name="Priya", salary=75000))
````

combining *args and **kwargs.

````python
def employee_summary(*args, **kwargs):
    return args, kwargs

print(employee_summary("Arun", 60000, department="HR", experience=5)) ## arun 6000is args and other are kwargs
````

````python
def record_employee(*args,**kwargs):
  return args,kwargs

print(record_employee("Priya", 75000, department="Analytics", city="Bengaluru"))
````

Variable scope


Variable scope means where a variable can be accessed in your Python code.


Local variable

A variable created inside a function is usually local to that function.


Global variable

A variable created outside a function can generally be read inside it.

````python
def calculate_salary():
    salary = 60000     ## local to this function
    return salary

print(calculate_salary())  # 60000
print(salary)              # NameError
````

````python
company = "Shell"     ### global

def show_company():
    print(company)

show_company()
````

````python
company = "Shell"

def update_company():
    company = "Google"     ### dosnt change global var
    return company

print(update_company())
print(company)
````

````python
company = "Shell"

def update_company():
    global company       ## if we want to change the global var
    company = "Google"

update_company()
print(company)
````

````python
def analyze_sales(sales):
    total = sum(sales)
    highest = max(sales)
    lowest = min(sales)

    return total, highest, lowest

result = analyze_sales([1000, 2500, 800, 3000]) ### see input is list we can
##also use tuples
print(result)
````

Calling one function inside another


In Python, one function can call another function. This is useful when you want to break a larger task into smaller, reusable steps.

For example, imagine calculating the total salary and then calculating a bonus based on that salary.

````python
def calculate_total_salary(basic, allowance):
    return basic + allowance

def calculate_bonus(basic, allowance):
    total = calculate_total_salary(basic, allowance)
    return total * 0.10

bonus = calculate_bonus(50000, 10000)
print(bonus)
````

````python
##calculate_total(price, quantity) — returns the total cost.
def Calculate_total(price,quantity):
  Total_cost = price * quantity
  return Total_cost

##calculate_discounted_price(price, quantity, discount_percent) —
##calls the first function to calculate the total, then subtracts the
##specified percentage discount.

def calculate_discounted_price(price, quantity, discount_percent):
    total = calculate_total(price, quantity)
    discount = total * discount_percent / 100
    return total - discount

print(Calculate_discounted_price(1000,3,10))

````

##Exception handling:
Sometimes a function receives unexpected input and causes an error. Exception handling lets us handle certain errors gracefully instead of stopping the program unexpectedly.




````python
def calculate_average(total, count):
    try:
        return total / count   ### if this throws an error then enter Expect
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(calculate_average(1000, 0))
print(calculate_average(1000, 4))
````

````python
def calculate_discounted_price(price, discount):
  try:
    required_result = price - discount
    return required_result
  except TypeError:
    return "Some error in typing"

##print(calculate_discounted_price(1000, 200)) ## gives 800
print(calculate_discounted_price(1000, "two hundred"))
````
